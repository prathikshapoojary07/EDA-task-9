```python
print("Zero values in bedrooms:",(df["bedrooms"]==0).sum())
print("Zero values in bathrooms:",(df["bathrooms"]==0).sum())
```

```python
df["bedrooms"]=df["bedrooms"].replace(0, np.nan)
df["bathrooms"]=df["bathrooms"].replace(0, np.nan)
```

```python
df.isnull().sum()
```

```python
df["bedrooms"].mean()         
```

```python
df["bedrooms"]=df["bedrooms"].fillna(df["bedrooms"].mean())
```

```python
df["bathrooms"]=df["bathrooms"].fillna(df["bathrooms"].mean())
```

```python
df.isnull().sum()
```

```python
df["date"] = pd.to_datetime(df["date"])
```

```python
df["sale_year"] = df["date"].dt.year
```

```python
df["house_age"] = df["sale_year"] - df["yr_built"]
```

```python
df["years_since_renovation"] = np.where(
    df["yr_renovated"] > 0,
    df["sale_year"] - df["yr_renovated"],
    np.nan
)
```

```python
df[["date", "sale_year", "yr_built", "yr_renovated",
    "house_age", "years_since_renovation"]].head()
```

```python
df["price_per_sqft_living"] = df["price"] / df["sqft_living"]
df["price_per_sqft_lot"] = df["price"] / df["sqft_lot"]
```

```python
df[["price", "sqft_living", "sqft_lot",
    "price_per_sqft_living", "price_per_sqft_lot"]].head()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x=df["price"])
plt.title("Box plot of house price")
plt.xlabel("Price")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x=df["price_per_sqft_living"])
plt.title("Box plot of price_per_sqft_living")
plt.xlabel("Price per sqft living")
plt.show()
```

```python
Q1=df["price"].quantile(0.25)
Q3=df["price"].quantile(0.75)

IQR=Q3-Q1

lower_bound=Q1-1.5*IQR
upper_bound=Q3+1.5*IQR

print("Q1:",Q1)
print("Q3:",Q3)
print("IQR:",IQR)
print("Lower Bound:",lower_bound)
print("Upper Bound:",upper_bound)
```

```python
print_outlier=df[
    (df["price"]<lower_bound) |
    (df["price"]>upper_bound)
]

print("Number of Price Outlier:",len(print_outlier))

print_outlier[["id","price","bedrooms",
                "bathrooms","sqft_living"]].head(20)
```
